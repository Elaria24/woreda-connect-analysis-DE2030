```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

cluster_results = []

for k in range(2, 10):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)

    inertia = kmeans.inertia_

    silhouette = silhouette_score(
        X_scaled,
        labels
    )

    cluster_results.append(
        {
            "K": k,
            "Inertia": inertia,
            "Silhouette": silhouette
        }
    )

import pandas as pd

cluster_results = pd.DataFrame(cluster_results)

cluster_results
```

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans

# ==========================================================
# FEATURE ENGINEERING
# ==========================================================

# Population served per LTE tower (replace 0 towers with 1 to avoid division by zero)
final_gdf["population_per_LTE"] = (
    final_gdf["population"] /
    final_gdf["LTE_tower_count"].replace(0, 1)
)

# Binary indicators
final_gdf["has_LTE"] = (final_gdf["LTE_tower_count"] > 0).astype(int)
final_gdf["has_fiber"] = (final_gdf["fiber_length_km"] > 0).astype(int)
final_gdf["has_energy"] = (final_gdf["energy_length_km"] > 0).astype(int)

# ==========================================================
# NORMALIZATION
# ==========================================================

features = [
    "population",
    "Total_Economic_Assets",
    "LTE_tower_count",
    "fiber_length_km",
    "energy_length_km",
    "population_per_LTE"
]

scaler = MinMaxScaler()

scaled = pd.DataFrame(
    scaler.fit_transform(final_gdf[features]),
    columns=features,
    index=final_gdf.index
)

# ==========================================================
# KMEANS (k=4)
# ==========================================================

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=20
)

final_gdf["cluster"] = kmeans.fit_predict(scaled)

# ==========================================================
# DEMAND SCORE
# ==========================================================

final_gdf["demand_score"] = (
    0.60 * scaled["population"] +
    0.40 * scaled["Total_Economic_Assets"]
)

# ==========================================================
# INFRASTRUCTURE SCORE
# ==========================================================

final_gdf["infrastructure_score"] = (
    0.45 * scaled["LTE_tower_count"] +
    0.30 * scaled["fiber_length_km"] +
    0.25 * scaled["energy_length_km"]
)

# ==========================================================
# DIGITAL GAP
# ==========================================================

final_gdf["digital_gap"] = (
    final_gdf["demand_score"] -
    final_gdf["infrastructure_score"]
)

# ==========================================================
# PRIORITY SCORE
# ==========================================================

final_gdf["priority_score"] = (
    0.65 * final_gdf["digital_gap"] +
    0.35 * scaled["population_per_LTE"]
)

# Normalize to 0-1
min_score = final_gdf["priority_score"].min()
max_score = final_gdf["priority_score"].max()

final_gdf["priority_score"] = (
    (final_gdf["priority_score"] - min_score)
    / (max_score - min_score)
)

print("Done!")
```

```python
conditions = [

    (
        (final_gdf["LTE_tower_count"] == 0) &
        (final_gdf["fiber_length_km"] > 0) &
        (final_gdf["energy_length_km"] > 0)
    ),

    (
        (final_gdf["LTE_tower_count"] == 0)
    ),

    (
        (final_gdf["population_per_LTE"] > 50000)
    ),

    (
        (final_gdf["fiber_length_km"] == 0)
    )

]

choices = [

    "Deploy LTE Tower",

    "Build New Digital Infrastructure",

    "Expand LTE Capacity",

    "Extend Fiber Network"

]

final_gdf["recommendation"] = np.select(
    conditions,
    choices,
    default="Maintain Current Infrastructure"
)
```

```python
import numpy as np

# Pairwise comparison matrix
A = np.array([
    [1,   2,   1/3, 1/2, 2],    # Population
    [1/2, 1,   1/4, 1/3, 1],    # Assets
    [3,   4,   1,   2,   3],    # LTE
    [2,   3,   1/2, 1,   2],    # Fiber
    [1/2, 1,   1/3, 1/2, 1]     # Energy
])

# Principal eigenvector
eigenvalues, eigenvectors = np.linalg.eig(A)

max_index = np.argmax(eigenvalues.real)

weights = eigenvectors[:, max_index].real
weights = weights / weights.sum()

criteria = [
    "population",
    "assets",
    "LTE",
    "fiber",
    "energy"
]

for c, w in zip(criteria, weights):
    print(f"{c:12s}: {w:.3f}")
```

```python
n = A.shape[0]

lambda_max = eigenvalues[max_index].real

CI = (lambda_max - n) / (n - 1)

RI = 1.12   # Random Index for n = 5

CR = CI / RI

print("Consistency Ratio:", round(CR,3))

if CR < 0.10:
    print("Pairwise comparisons are acceptably consistent.")
else:
    print("Review the pairwise comparisons; consistency is low.")
```

```python
final_gdf["readiness_index"] = (
      weights[2] * scaled["LTE_tower_count"]
    + weights[3] * scaled["fiber_length_km"]
    + weights[4] * scaled["energy_length_km"]
)

final_gdf["demand_index"] = (
      weights[0] * scaled["population"]
    + weights[1] * scaled["Total_Economic_Assets"]
)

final_gdf["priority_score"] = (
    final_gdf["demand_index"] -
    final_gdf["readiness_index"]
)

# Normalize to 0–1
final_gdf["priority_score"] = (
    (final_gdf["priority_score"] - final_gdf["priority_score"].min())
    /
    (final_gdf["priority_score"].max() - final_gdf["priority_score"].min())
)
```

```python
# ==========================================================
# EXPLAINABLE RECOMMENDATION ENGINE
# ==========================================================

reasons = []
recommendations = []

# Data-driven thresholds
pop75 = final_gdf["population"].quantile(0.75)
asset75 = final_gdf["Total_Economic_Assets"].quantile(0.75)
pressure75 = final_gdf["population_per_LTE"].quantile(0.75)
fiber25 = final_gdf["fiber_length_km"].quantile(0.25)
energy25 = final_gdf["energy_length_km"].quantile(0.25)

for _, row in final_gdf.iterrows():

    reason = []

    # -----------------------------
    # DEMAND
    # -----------------------------
    if row["population"] >= pop75:
        reason.append("High population")

    if row["Total_Economic_Assets"] >= asset75:
        reason.append("High economic activity")

    # -----------------------------
    # LTE
    # -----------------------------
    if row["LTE_tower_count"] == 0:
        reason.append("No LTE towers")

    elif row["population_per_LTE"] >= pressure75:
        reason.append("Existing LTE overloaded")

    # -----------------------------
    # Fiber
    # -----------------------------
    if row["fiber_length_km"] > fiber25:
        reason.append("Fiber available")
    else:
        reason.append("Limited fiber")

    # -----------------------------
    # Energy
    # -----------------------------
    if row["energy_length_km"] > energy25:
        reason.append("Energy available")
    else:
        reason.append("Limited energy")

    # -----------------------------
    # FINAL RECOMMENDATION
    # -----------------------------
    if (
        row["LTE_tower_count"] == 0
        and row["population"] >= pop75
        and (
            row["fiber_length_km"] > fiber25
            or row["energy_length_km"] > energy25
        )
    ):

        rec = "Deploy LTE Tower"

    elif (
        row["LTE_tower_count"] == 0
        and row["population"] >= pop75
    ):

        rec = "Build Digital Infrastructure"

    elif (
        row["population_per_LTE"] >= pressure75
    ):

        rec = "Expand LTE Capacity"

    elif (
        row["fiber_length_km"] <= fiber25
        and row["LTE_tower_count"] > 0
    ):

        rec = "Extend Fiber Network"

    elif (
        row["energy_length_km"] <= energy25
        and row["LTE_tower_count"] > 0
    ):

        rec = "Improve Power Infrastructure"

    else:

        rec = "Maintain and Monitor"

    recommendations.append(rec)
    reasons.append("; ".join(reason))

# Save results
final_gdf["recommendation"] = recommendations
final_gdf["reason"] = reasons

# ==========================================================
# FINAL DASHBOARD TABLE
# ==========================================================

dashboard_df = final_gdf[
    [
        "REGIONNAME",
        "ZONENAME",
        "WOREDANAME",
        "cluster",
        "priority_score",
        "recommendation",
        "reason",
        "population",
        "LTE_tower_count",
        "fiber_length_km",
        "energy_length_km",
        "Total_Economic_Assets"
    ]
].sort_values(
    "priority_score",
    ascending=False
)

print("="*80)
print("TOP 25 PRIORITY WOREDAS")
print("="*80)

display(dashboard_df.head(25))

print("\nRecommendation Summary\n")
print(final_gdf["recommendation"].value_counts())

print("\nCluster Summary\n")
print(final_gdf["cluster"].value_counts())

print("\nProject pipeline completed successfully.")
```
